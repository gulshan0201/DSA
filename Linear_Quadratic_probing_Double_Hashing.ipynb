```python
#Quadratic Probing using hashing
class HashTableQuadraticProbing:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def __hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.__hash(key)
        i = 0
        while i < self.size:
            probe_index = (index + i**2) % self.size
            if self.table[probe_index] is None:
                self.table[probe_index] = (key, value)
                return True
            i += 1
        return False  # Table is full

    def search(self, key):
        index = self.__hash(key)
        i = 0
        while i < self.size:
            probe_index = (index + i**2) % self.size
            if self.table[probe_index] is None:
                return None  # Key not found
            if self.table[probe_index][0] == key:
                return self.table[probe_index][1]  # Key found
            i += 1
        return None  # Key not found

    def delete(self, key):
        index = self.__hash(key)
        i = 0
        while i < self.size:
            probe_index = (index + i**2) % self.size
            if self.table[probe_index] is None:
                return False  # Key not found
            if self.table[probe_index][0] == key:
                self.table[probe_index] = None  # Mark as deleted
                return True
            i += 1
        return False  # Key not found

    def display(self):
        print(self.table)


hash_table_qp = HashTableQuadraticProbing(10)

# Insert some key-value pairs
hash_table_qp.insert("apple", 1)
hash_table_qp.insert("banana", 2)
hash_table_qp.insert("orange", 3)
hash_table_qp.insert("grape", 4)
hash_table_qp.insert("melon", 5)

# Display the hash table
print("Hash Table after insertions (Quadratic Probing):")
hash_table_qp.display()

# Search for a key
print("\nSearching for 'banana':", hash_table_qp.search("banana"))
print("Searching for 'kiwi':", hash_table_qp.search("kiwi"))

# Delete a key
print("\nDeleting 'orange':", hash_table_qp.delete("orange"))
print("Hash Table after deletion (Quadratic Probing):")
hash_table_qp.display()

# Search for the deleted key
print("\nSearching for 'orange' after deletion:", hash_table_qp.search("orange"))
```

```python
class HashTableLinear:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)

        for i in range(self.size):  # Linear Probing
            probe_index = (index + i) % self.size
            if self.table[probe_index] is None:
                self.table[probe_index] = (key, value)
                return True
        return False  # Table full

    def search(self, key):
        index = self.hash_function(key)

        for i in range(self.size):  # Linear Probing
            probe_index = (index + i) % self.size
            if self.table[probe_index] is None:
                return None
            if self.table[probe_index][0] == key:
                return self.table[probe_index][1]
        return None

    def delete(self, key):
        index = self.hash_function(key)

        for i in range(self.size):  # Linear Probing
            probe_index = (index + i) % self.size
            if self.table[probe_index] is None:
                return False
            if self.table[probe_index][0] == key:
                self.table[probe_index] = None
                return True
        return False

    def display(self):
        print(self.table)





# inputs
print("=== Linear Probing ===")
ht_linear = HashTableLinear(10)
ht_linear.insert("apple", 1)
ht_linear.insert("banana", 2)
ht_linear.insert("orange", 3)
ht_linear.display()
print("Search banana:", ht_linear.search("banana"))
print("Delete orange:", ht_linear.delete("orange"))
ht_linear.display()
```

```python
class HashTableDouble:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash1(self, key):
        return hash(key) % self.size

    def hash2(self, key):
        prime = 7  # should be < size and coprime
        return prime - (hash(key) % prime)

    def insert(self, key, value):
        index = self.hash1(key)
        step = self.hash2(key)

        for i in range(self.size):  # Double Hashing
            probe_index = (index + i * step) % self.size
            if self.table[probe_index] is None:
                self.table[probe_index] = (key, value)
                return True
        return False

    def search(self, key):
        index = self.hash1(key)
        step = self.hash2(key)

        for i in range(self.size):  # Double Hashing
            probe_index = (index + i * step) % self.size
            if self.table[probe_index] is None:
                return None
            if self.table[probe_index][0] == key:
                return self.table[probe_index][1]
        return None

    def delete(self, key):
        index = self.hash1(key)
        step = self.hash2(key)

        for i in range(self.size):  # Double Hashing
            probe_index = (index + i * step) % self.size
            if self.table[probe_index] is None:
                return False
            if self.table[probe_index][0] == key:
                self.table[probe_index] = None
                return True
        return False

    def display(self):
        print(self.table)


print("\n=== Double Hashing ===")
ht_double = HashTableDouble(10)
ht_double.insert("apple", 1)
ht_double.insert("banana", 2)
ht_double.insert("orange", 3)
ht_double.display()
print("Search banana:", ht_double.search("banana"))
print("Delete orange:", ht_double.delete("orange"))
ht_double.display()
```
