```python
def climb_stairs(n):
    # Base cases: If there are 0 or 1 steps, there's only 1 way
    if n <= 1:
        return 1

    # Create a list to store the number of ways for each step
    # We need n + 1 size to include the nth step
    dp = [0] * (n + 1)

    # Starting points
    dp[0] = 1  # 1 way to stay at the ground
    dp[1] = 1  # 1 way to reach the first step

    # Fill the table for all steps from 2 to n
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]

# Test it
steps = 5
print(f"Total ways to climb {steps} steps: {climb_stairs(steps)}")
# Output: 8
```
